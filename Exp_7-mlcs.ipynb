```python
# Experiment 7: Build a machine learning model to solve CAPTCHA challenges.
# ============================================================
# MACHINE LEARNING FOR CAPTCHA CHARACTER RECOGNITION
# Cybersecurity / ML Academic Project
#
# Dataset:
# Synthetic CAPTCHA images generated by this notebook.
#
# Model:
# CNN (Convolutional Neural Network)
#
# IMPORTANT:
# This project is for educational research using synthetic/
# authorized CAPTCHA data. Do not use it to bypass CAPTCHAs
# on third-party websites.
# ============================================================


# ============================================================
# CELL 1 - INSTALL REQUIRED LIBRARIES
# ============================================================

!pip install tensorflow pillow numpy matplotlib scikit-learn -q


# ============================================================
# CELL 2 - IMPORT LIBRARIES
# ============================================================

import os
import random
import string

import numpy as np
import matplotlib.pyplot as plt

from PIL import Image, ImageDraw, ImageFont

import tensorflow as tf

from tensorflow.keras import layers, models
from sklearn.model_selection import train_test_split

print("TensorFlow version:", tf.__version__)


# ============================================================
# CELL 3 - SET RANDOM SEEDS
# ============================================================

# Setting seeds makes the experiment more reproducible.

random.seed(42)
np.random.seed(42)
tf.random.set_seed(42)


# ============================================================
# CELL 4 - CAPTCHA SETTINGS
# ============================================================

# Number of CAPTCHA images to generate
TOTAL_IMAGES = 3000

# Number of characters in each CAPTCHA
CAPTCHA_LENGTH = 5

# Image dimensions
IMG_WIDTH = 160
IMG_HEIGHT = 60

# Characters that can appear in the CAPTCHA
CHARACTERS = string.ascii_uppercase + string.digits

# Number of possible characters
NUM_CLASSES = len(CHARACTERS)

print("Characters:", CHARACTERS)
print("Number of classes:", NUM_CLASSES)


# ============================================================
# CELL 5 - CREATE DATASET FOLDER
# ============================================================

DATASET_PATH = "synthetic_captcha_dataset"

os.makedirs(DATASET_PATH, exist_ok=True)

print("Dataset folder:", DATASET_PATH)


# ============================================================
# CELL 6 - FIND A FONT
# ============================================================

# The CAPTCHA generator needs a font.
# We try common locations automatically.

possible_fonts = [
    "/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf",
    "/usr/share/fonts/truetype/liberation2/LiberationSans-Bold.ttf",
    "C:/Windows/Fonts/arialbd.ttf",
    "C:/Windows/Fonts/Arial.ttf"
]

FONT_PATH = None

for font in possible_fonts:
    if os.path.exists(font):
        FONT_PATH = font
        break

if FONT_PATH is None:
    raise FileNotFoundError(
        "No suitable font found. Please change FONT_PATH manually."
    )

print("Using font:", FONT_PATH)


# ============================================================
# CELL 7 - FUNCTION TO GENERATE ONE CAPTCHA
# ============================================================

def generate_captcha(text, filename):
    """
    Creates one synthetic CAPTCHA image.

    The image contains:
    - Random background
    - CAPTCHA characters
    - Random lines
    - Random noise
    - Small character rotations
    """

    # Create a light background
    image = Image.new(
        "RGB",
        (IMG_WIDTH, IMG_HEIGHT),
        (245, 245, 245)
    )

    draw = ImageDraw.Draw(image)

    # Font size
    font = ImageFont.truetype(FONT_PATH, 38)

    # Width available for each character
    char_width = IMG_WIDTH // CAPTCHA_LENGTH

    # --------------------------------------------------------
    # Draw characters
    # --------------------------------------------------------

    for i, char in enumerate(text):

        # Create a temporary transparent image
        char_image = Image.new(
            "RGBA",
            (50, 60),
            (255, 255, 255, 0)
        )

        char_draw = ImageDraw.Draw(char_image)

        # Random dark color
        char_color = (
            random.randint(0, 80),
            random.randint(0, 80),
            random.randint(0, 80),
            255
        )

        # Draw character
        char_draw.text(
            (5, 5),
            char,
            font=font,
            fill=char_color
        )

        # Random rotation makes recognition more difficult
        angle = random.randint(-15, 15)

        char_image = char_image.rotate(
            angle,
            resample=Image.Resampling.BICUBIC,
            expand=1
        )

        # Character position
        x = i * char_width + 5
        y = random.randint(-5, 5)

        image.paste(
            char_image,
            (x, y),
            char_image
        )

    # --------------------------------------------------------
    # Add random interference lines
    # --------------------------------------------------------

    for _ in range(5):

        x1 = random.randint(0, IMG_WIDTH)
        y1 = random.randint(0, IMG_HEIGHT)

        x2 = random.randint(0, IMG_WIDTH)
        y2 = random.randint(0, IMG_HEIGHT)

        line_color = (
            random.randint(80, 180),
            random.randint(80, 180),
            random.randint(80, 180)
        )

        draw.line(
            (x1, y1, x2, y2),
            fill=line_color,
            width=random.randint(1, 2)
        )

    # --------------------------------------------------------
    # Add random noise dots
    # --------------------------------------------------------

    for _ in range(250):

        x = random.randint(0, IMG_WIDTH - 1)
        y = random.randint(0, IMG_HEIGHT - 1)

        noise_color = (
            random.randint(100, 220),
            random.randint(100, 220),
            random.randint(100, 220)
        )

        draw.point(
            (x, y),
            fill=noise_color
        )

    # Save image
    image.save(filename)


# ============================================================
# CELL 8 - GENERATE SYNTHETIC CAPTCHA DATASET
# ============================================================

print("Generating synthetic CAPTCHA dataset...")

for i in range(TOTAL_IMAGES):

    # Generate random CAPTCHA text
    captcha_text = "".join(
        random.choice(CHARACTERS)
        for _ in range(CAPTCHA_LENGTH)
    )

    # Filename contains the correct label
    filename = os.path.join(
        DATASET_PATH,
        captcha_text + "_" + str(i) + ".png"
    )

    generate_captcha(
        captcha_text,
        filename
    )

print("Dataset generation completed.")


# ============================================================
# CELL 9 - DISPLAY SAMPLE CAPTCHAS
# ============================================================

files = os.listdir(DATASET_PATH)

plt.figure(figsize=(12, 6))

for i in range(10):

    filename = files[i]

    image_path = os.path.join(
        DATASET_PATH,
        filename
    )

    image = Image.open(image_path)

    plt.subplot(2, 5, i + 1)

    plt.imshow(image)

    # Extract CAPTCHA text from filename
    label = filename.split("_")[0]

    plt.title(label)

    plt.axis("off")

plt.tight_layout()
plt.show()


# ============================================================
# CELL 10 - LOAD IMAGES AND LABELS
# ============================================================

images = []
labels = []

for filename in files:

    if not filename.endswith(".png"):
        continue

    # Extract CAPTCHA text from filename
    captcha_text = filename.split("_")[0]

    # Load image
    image_path = os.path.join(
        DATASET_PATH,
        filename
    )

    image = Image.open(
        image_path
    ).convert("L")

    # Resize image
    image = image.resize(
        (IMG_WIDTH, IMG_HEIGHT)
    )

    # Convert image to NumPy array
    image = np.array(image)

    # Normalize pixel values from 0-255 to 0-1
    image = image.astype("float32") / 255.0

    images.append(image)
    labels.append(captcha_text)


# Convert lists to NumPy arrays
images = np.array(images)
labels = np.array(labels)

# CNN expects:
# height x width x channels

images = images[..., np.newaxis]

print("Image dataset shape:", images.shape)
print("Label dataset shape:", labels.shape)


# ============================================================
# CELL 11 - CREATE CHARACTER ENCODING
# ============================================================

# Convert characters to numbers.

char_to_num = {
    char: i
    for i, char in enumerate(CHARACTERS)
}

# Convert numbers back to characters.

num_to_char = {
    i: char
    for char, i in char_to_num.items()
}


# ============================================================
# CELL 12 - ENCODE CAPTCHA LABELS
# ============================================================

encoded_labels = np.zeros(
    (len(labels), CAPTCHA_LENGTH),
    dtype=np.int32
)

for i, label in enumerate(labels):

    for j, char in enumerate(label):

        encoded_labels[i, j] = char_to_num[char]


print("Encoded labels shape:", encoded_labels.shape)

print("\nExample:")

print("Original:", labels[0])

print("Encoded :", encoded_labels[0])


# ============================================================
# CELL 13 - TRAIN/TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    images,
    encoded_labels,
    test_size=0.20,
    random_state=42
)

print("Training images:", X_train.shape)
print("Testing images :", X_test.shape)


# ============================================================
# CELL 14 - PREPARE LABELS FOR MULTI-OUTPUT CNN
# ============================================================

# Our CNN has one output for each CAPTCHA character.
#
# Example:
#
# CAPTCHA = A7K3P
#
# Output 1 -> A
# Output 2 -> 7
# Output 3 -> K
# Output 4 -> 3
# Output 5 -> P

y_train_list = [
    y_train[:, i]
    for i in range(CAPTCHA_LENGTH)
]

y_test_list = [
    y_test[:, i]
    for i in range(CAPTCHA_LENGTH)
]


# ============================================================
# CELL 15 - BUILD CNN MODEL
# ============================================================

input_layer = layers.Input(
    shape=(IMG_HEIGHT, IMG_WIDTH, 1)
)


# First convolution layer
x = layers.Conv2D(
    32,
    (3, 3),
    activation="relu",
    padding="same"
)(input_layer)

x = layers.MaxPooling2D(
    (2, 2)
)(x)


# Second convolution layer
x = layers.Conv2D(
    64,
    (3, 3),
    activation="relu",
    padding="same"
)(x)

x = layers.MaxPooling2D(
    (2, 2)
)(x)


# Third convolution layer
x = layers.Conv2D(
    128,
    (3, 3),
    activation="relu",
    padding="same"
)(x)

x = layers.MaxPooling2D(
    (2, 2)
)(x)


# Flatten extracted features
x = layers.Flatten()(x)


# Fully connected layer
x = layers.Dense(
    256,
    activation="relu"
)(x)


# Dropout helps reduce overfitting
x = layers.Dropout(
    0.4
)(x)


# ============================================================
# CREATE ONE OUTPUT FOR EACH CHARACTER
# ============================================================

outputs = []

for i in range(CAPTCHA_LENGTH):

    output = layers.Dense(
        NUM_CLASSES,
        activation="softmax",
        name=f"character_{i+1}"
    )(x)

    outputs.append(output)


# Create complete model
model = models.Model(
    inputs=input_layer,
    outputs=outputs
)


# Display model architecture
model.summary()


# ============================================================
# CELL 16 - COMPILE MODEL
# ============================================================

model.compile(

    optimizer="adam",

    # One loss function for every character position
    loss=[
        "sparse_categorical_crossentropy"
        for _ in range(CAPTCHA_LENGTH)
    ],

    # Accuracy for every character position
    metrics=[
        "accuracy"
        for _ in range(CAPTCHA_LENGTH)
    ]
)

print("Model compiled successfully.")


# ============================================================
# CELL 17 - TRAIN THE CNN
# ============================================================

history = model.fit(

    X_train,

    y_train_list,

    validation_split=0.20,

    epochs=15,

    batch_size=32,

    verbose=1
)


# ============================================================
# CELL 18 - PLOT TRAINING LOSS
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.title(
    "CNN Training and Validation Loss"
)

plt.legend()

plt.show()


# ============================================================
# CELL 19 - EVALUATE MODEL
# ============================================================

results = model.evaluate(
    X_test,
    y_test_list,
    verbose=1
)

print("\nModel evaluation completed.")


# ============================================================
# CELL 20 - MAKE PREDICTIONS
# ============================================================

predictions = model.predict(
    X_test,
    verbose=0
)

print("Predictions generated.")


# ============================================================
# CELL 21 - CALCULATE CHARACTER-LEVEL ACCURACY
# ============================================================

predicted_matrix = np.zeros_like(
    y_test
)

print("\nCharacter Accuracy:")

for i in range(CAPTCHA_LENGTH):

    # Select predicted class
    predicted_matrix[:, i] = np.argmax(
        predictions[i],
        axis=1
    )

    # Calculate accuracy
    accuracy = np.mean(
        predicted_matrix[:, i] == y_test[:, i]
    )

    print(
        f"Character {i+1}: "
        f"{accuracy * 100:.2f}%"
    )


# ============================================================
# CELL 22 - COMPLETE CAPTCHA ACCURACY
# ============================================================

# A CAPTCHA is considered completely correct
# only when ALL characters are predicted correctly.

correct_captchas = np.all(
    predicted_matrix == y_test,
    axis=1
)

captcha_accuracy = np.mean(
    correct_captchas
)

print(
    "\nComplete CAPTCHA Accuracy:",
    f"{captcha_accuracy * 100:.2f}%"
)


# ============================================================
# CELL 23 - SHOW ACTUAL VS PREDICTED CAPTCHA
# ============================================================

plt.figure(figsize=(12, 8))

number_of_examples = min(
    10,
    len(X_test)
)

for i in range(number_of_examples):

    # Build predicted CAPTCHA
    predicted_text = ""

    for j in range(CAPTCHA_LENGTH):

        predicted_index = np.argmax(
            predictions[j][i]
        )

        predicted_text += num_to_char[
            predicted_index
        ]


    # Build actual CAPTCHA
    actual_text = ""

    for j in range(CAPTCHA_LENGTH):

        actual_text += num_to_char[
            y_test[i][j]
        ]


    # Display image
    plt.subplot(2, 5, i + 1)

    plt.imshow(
        X_test[i].squeeze(),
        cmap="gray"
    )

    plt.title(
        f"Actual: {actual_text}\n"
        f"Predicted: {predicted_text}"
    )

    plt.axis("off")


plt.tight_layout()
plt.show()


# ============================================================
# CELL 24 - TEST ONE CAPTCHA
# ============================================================

sample_index = 0

sample_image = X_test[
    sample_index
]


# Predict the sample
sample_prediction = model.predict(
    sample_image[np.newaxis, ...],
    verbose=0
)


# Convert predictions into characters
predicted_text = ""

for i in range(CAPTCHA_LENGTH):

    predicted_index = np.argmax(
        sample_prediction[i][0]
    )

    predicted_text += num_to_char[
        predicted_index
    ]


# Actual CAPTCHA
actual_text = ""

for i in range(CAPTCHA_LENGTH):

    actual_text += num_to_char[
        y_test[sample_index][i]
    ]


print("Actual CAPTCHA   :", actual_text)
print("Predicted CAPTCHA:", predicted_text)


# ============================================================
# CELL 25 - DISPLAY FINAL TEST RESULT
# ============================================================

plt.figure(figsize=(8, 3))

plt.imshow(
    sample_image.squeeze(),
    cmap="gray"
)

plt.title(
    f"Actual: {actual_text} | "
    f"Predicted: {predicted_text}"
)

plt.axis("off")

plt.show()


# ============================================================
# FINAL PROJECT RESULTS
# ============================================================

print("=" * 55)
print("CAPTCHA MACHINE LEARNING EXPERIMENT COMPLETED")
print("=" * 55)

print("Total synthetic images :", TOTAL_IMAGES)
print("Training images        :", len(X_train))
print("Testing images         :", len(X_test))
print("CAPTCHA length         :", CAPTCHA_LENGTH)
print("Number of classes      :", NUM_CLASSES)

print(
    "Complete CAPTCHA Accuracy:",
    f"{captcha_accuracy * 100:.2f}%"
)

print("=" * 55)
```
